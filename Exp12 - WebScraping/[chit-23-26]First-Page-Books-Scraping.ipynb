# CHIT No. 23 — Scrape Books from the First Page

Scrapes each book's title, price, rating and availability from the first page of books.toscrape.com and stores the results in a pandas dataframe.

**How to use:** Run each cell from top to bottom. Read the short explanation before every code cell.

### What this cell does
Import requests to download the page, BeautifulSoup to read HTML, and pandas to create a dataframe.

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

### What this cell does
Set the first-page URL and download its HTML. Raise an error if the request fails.

```python
url = "https://books.toscrape.com/"
response = requests.get(url, timeout=20)
response.raise_for_status()
print("Page downloaded successfully:", response.status_code)
```

### What this cell does
Parse the HTML and find each book card. Each article.product_pod contains one book.

```python
soup = BeautifulSoup(response.text, "html.parser")
books = soup.select("article.product_pod")
print("Number of books found:", len(books))
```

### What this cell does
Loop through the book cards and extract title, price, star rating and availability text.

```python
rows = []

rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

for book in books:
    title = book.h3.a["title"]
    price = book.select_one(".price_color").get_text(strip=True)
    rating_word = book.select_one("p.star-rating")["class"][1]
    rating = rating_map.get(rating_word, None)
    availability = book.select_one(".availability").get_text(" ", strip=True)

    rows.append({
        "Title": title,
        "Price": price,
        "Rating": rating,
        "Availability": availability
    })

df = pd.DataFrame(rows)
df.head()
```

### What this cell does
Display the complete dataframe containing all books from the first page.

```python
print("Total books:", len(df))
display(df)
```
