# CHIT No. 25 — Scrape Quotes from Two Pages

Scrapes quotes, authors and tags from the first two pages of quotes.toscrape.com and displays the number of quotes written by each author.

**How to use:** Run each cell from top to bottom. Read the short explanation before every code cell.

### What this cell does
Import the required libraries.

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

### What this cell does
Create an empty list for the quotes collected from both pages.

```python
all_quotes = []
```

### What this cell does
Visit pages 1 and 2. For each quote block, extract its text, author and all tags.

```python
for page_num in range(1, 3):
    url = f"https://quotes.toscrape.com/page/{page_num}/"
    response = requests.get(url, timeout=20)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, "html.parser")

    quote_blocks = soup.select("div.quote")
    for item in quote_blocks:
        quote_text = item.select_one("span.text").get_text(strip=True)
        author = item.select_one("small.author").get_text(strip=True)
        tags = [tag.get_text(strip=True) for tag in item.select("a.tag")]

        all_quotes.append({
            "Quote": quote_text,
            "Author": author,
            "Tags": ", ".join(tags),
            "Page": page_num
        })

    print(f"Page {page_num}: {len(quote_blocks)} quotes found")
```

### What this cell does
Create a dataframe and display the scraped quotes, authors and tags.

```python
df = pd.DataFrame(all_quotes)
display(df)
```

### What this cell does
Count the number of quotes for each author and sort from most to fewest.

```python
author_counts = df["Author"].value_counts().rename_axis("Author").reset_index(name="Number of Quotes")
display(author_counts)
```
