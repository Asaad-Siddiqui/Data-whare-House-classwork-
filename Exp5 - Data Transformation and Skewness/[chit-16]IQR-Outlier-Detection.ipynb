# CHIT No. 16 — IQR Method for Outlier Detection

This notebook finds outliers using the Interquartile Range (IQR) method, prints Q1, Q3, IQR and the fences, and draws a box plot.

**How to use:** Run the cells from top to bottom. Each code cell has a short explanation before it.

### What this cell does
Import pandas and seaborn/matplotlib for the table and box plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### What this cell does
Create a sample numeric column with a few unusually large values.

```python
data = pd.DataFrame({
    "Marks": [45, 48, 50, 52, 53, 54, 55, 56, 58, 60, 62, 150, 180]
})

data
```

### What this cell does
Calculate Q1 (25th percentile), Q3 (75th percentile), and IQR (Q3 minus Q1).

```python
Q1 = data["Marks"].quantile(0.25)
Q3 = data["Marks"].quantile(0.75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

### What this cell does
Calculate the lower and upper fences. Values outside these fences are considered outliers.

```python
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
```

### What this cell does
Select and display values that are below the lower fence or above the upper fence.

```python
outliers = data[(data["Marks"] < lower_fence) | (data["Marks"] > upper_fence)]

print("Outlier values:")
print(outliers["Marks"].tolist())
```

### What this cell does
Draw a box plot. Points beyond the whiskers are potential outliers.

```python
plt.figure(figsize=(7, 4))
sns.boxplot(x=data["Marks"])
plt.title("Box Plot of Marks")
plt.xlabel("Marks")
plt.show()
```
