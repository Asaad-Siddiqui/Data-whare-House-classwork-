# CHIT No. 15 — Data Transformation and Skewness

This notebook calculates skewness, applies log and square-root transformations, and compares histograms before and after. A small sample dataset is included, so no file upload is needed.

**How to use:** Run the cells from top to bottom. Each code cell has a short explanation before it.

### What this cell does
Import the libraries used for data, calculations, and charts.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### What this cell does
Create a numeric column with a right-skewed distribution. The larger values make the distribution skewed to the right.

```python
data = pd.DataFrame({
    "Income": [12000, 15000, 18000, 20000, 22000, 25000, 28000, 35000, 50000, 90000]
})

data
```

### What this cell does
Calculate and display the skewness of the original column. Positive skewness usually means a longer tail on the right.

```python
original_skewness = data["Income"].skew()
print("Original skewness:", round(original_skewness, 3))
```

### What this cell does
Apply log and square-root transformations. The log transformation uses log1p so it also works if values include zero.

```python
data["Log_Income"] = np.log1p(data["Income"])
data["Sqrt_Income"] = np.sqrt(data["Income"])

data
```

### What this cell does
Calculate skewness after each transformation so you can compare the results.

```python
print("Original skewness:", round(data["Income"].skew(), 3))
print("Log-transformed skewness:", round(data["Log_Income"].skew(), 3))
print("Square-root-transformed skewness:", round(data["Sqrt_Income"].skew(), 3))
```

### What this cell does
Draw histograms of the original values and both transformed columns. Compare how the shapes change.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(data["Income"], kde=True, ax=axes[0])
axes[0].set_title("Original Income")

sns.histplot(data["Log_Income"], kde=True, ax=axes[1])
axes[1].set_title("Log Transformation")

sns.histplot(data["Sqrt_Income"], kde=True, ax=axes[2])
axes[2].set_title("Square-root Transformation")

plt.tight_layout()
plt.show()
```

### What this cell does
Print a short conclusion using the skewness values. A value closer to zero indicates less skew, though the best transformation depends on the data.

```python
skews = {
    "Original": abs(data["Income"].skew()),
    "Log": abs(data["Log_Income"].skew()),
    "Square root": abs(data["Sqrt_Income"].skew())
}
print("Absolute skewness:", {k: round(v, 3) for k, v in skews.items()})
print("The transformation with the absolute skewness closest to zero is:",
      min(skews, key=skews.get))
```
