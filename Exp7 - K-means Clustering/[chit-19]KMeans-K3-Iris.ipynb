# CHIT No. 19 — K-means Clustering with K = 3

Uses the built-in Iris dataset, scales the features, applies K-means with 3 clusters, prints cluster sizes and centroids, and plots the clusters. No dataset download is required.

**How to use:** Run each cell from top to bottom. Read the short explanation before every code cell.

### What this cell does
Import the required libraries for data, scaling, clustering and plotting.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

### What this cell does
Load the Iris dataset and store its four numeric features in a dataframe.

```python
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
X.head()
```

### What this cell does
Scale the features so columns with larger numeric values do not dominate the clustering.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### What this cell does
Fit K-means with K = 3 and add the cluster label for each flower.

```python
model = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = model.fit_predict(X_scaled)
X["Cluster"] = labels

X.head()
```

### What this cell does
Print how many flowers belong to each cluster.

```python
print("Cluster sizes:")
print(X["Cluster"].value_counts().sort_index())
```

### What this cell does
Display the cluster centroids in scaled feature units.

```python
centroids = pd.DataFrame(model.cluster_centers_, columns=iris.feature_names)
print("Centroids (scaled feature values):")
print(centroids.round(3))
```

### What this cell does
Plot two Iris features using colours for clusters and mark the centroids with black X symbols.

```python
plt.figure(figsize=(8, 5))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap="viridis", s=45)
plt.scatter(model.cluster_centers_[:, 0], model.cluster_centers_[:, 1],
            marker="X", s=250, c="red", edgecolor="black", label="Centroids")
plt.xlabel(iris.feature_names[0] + " (scaled)")
plt.ylabel(iris.feature_names[1] + " (scaled)")
plt.title("Iris K-means Clustering (K = 3)")
plt.legend()
plt.show()
```
