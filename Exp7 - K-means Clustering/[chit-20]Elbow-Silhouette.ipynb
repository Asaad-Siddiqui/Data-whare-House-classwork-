# CHIT No. 20 — Elbow Curve and Silhouette Score

Plots the Elbow curve for K = 1 to 10 and calculates Silhouette scores for K = 2 to 6 using scaled Iris features. The notebook prints a suggested K and explains that the final choice should consider both plots and scores.

**How to use:** Run each cell from top to bottom. Read the short explanation before every code cell.

### What this cell does
Import libraries and load the Iris dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

### What this cell does
Scale the four Iris features before running K-means.

```python
iris = load_iris()
X = iris.data
X_scaled = StandardScaler().fit_transform(X)
```

### What this cell does
Fit K-means for K = 1 to 10 and save each model's inertia (within-cluster sum of squares).

```python
k_values = range(1, 11)
inertias = []

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    inertias.append(model.inertia_)

print("Inertia values:")
for k, inertia in zip(k_values, inertias):
    print(f"K={k}: {inertia:.2f}")
```

### What this cell does
Plot the Elbow curve. Look for the point where the decrease in inertia begins to slow down.

```python
plt.figure(figsize=(7, 4))
plt.plot(list(k_values), inertias, marker="o")
plt.xticks(list(k_values))
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Curve")
plt.grid(True)
plt.show()
```

### What this cell does
Calculate and print the Silhouette score for each K from 2 to 6. A higher score generally indicates better-separated clusters.

```python
silhouette_scores = {}

for k in range(2, 7):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores[k] = score
    print(f"K={k}: Silhouette score = {score:.3f}")
```

### What this cell does
Suggest the K with the highest Silhouette score and explain that the Elbow curve should also be checked before final selection.

```python
best_k = max(silhouette_scores, key=silhouette_scores.get)
print("Suggested K from Silhouette score:", best_k)
print("Justification: this K has the highest Silhouette score among K=2 to 6.")
print("Also inspect the Elbow curve and prefer a K where the inertia reduction starts to slow.")
```
