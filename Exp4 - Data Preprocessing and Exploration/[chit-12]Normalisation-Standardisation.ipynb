## Step 1: Create or load data\n\nWe scale two numeric columns: Age and Marks. Missing values are filled with the mean first.

> **What this cell does:** This creates sample student data. We use `Age` and `Marks` as the two numeric columns to scale.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.DataFrame({
    "Name": ["Amit", "Sara", "Rahul", "Priya", "Zoya", "John", "Maya", "Ali"],
    "Age": [20, 21, np.nan, 22, 20, 23, 21, 24],
    "Marks": [78, 85, 90, np.nan, 65, 95, 72, 88],
    "Attendance": [85, 90, 88, 92, np.nan, 96, 80, 89],
    "Department": ["Computer", "IT", "Computer", "IT", "Computer", "IT", "Computer", "IT"]
})
display(df)
```

## Step 2: Apply Min-Max and Z-score scaling\n\nMin-Max scales values from 0 to 1. Z-score standardisation scales values around 0.

> **What this cell does:** Missing values are filled with the mean first. Min-Max scaling changes values to a range from 0 to 1. Z-score standardisation expresses values as distances from the mean in standard-deviation units. The table compares the original and scaled values for five rows.


```python
columns = ["Age", "Marks"]
data = df[columns].copy().fillna(df[columns].mean())

minmax_values = MinMaxScaler().fit_transform(data)
zscore_values = StandardScaler().fit_transform(data)

result = pd.DataFrame({
    "Age_Original": data["Age"],
    "Age_MinMax": minmax_values[:, 0],
    "Age_Zscore": zscore_values[:, 0],
    "Marks_Original": data["Marks"],
    "Marks_MinMax": minmax_values[:, 1],
    "Marks_Zscore": zscore_values[:, 1]
})
display(result.head().round(3))
```

## Step 3: Plot histograms before and after scaling

> **What this cell does:** These histograms show the original values, Min-Max values, and Z-score values. Scaling changes the number range, while the data's general pattern remains similar.


```python
fig, axes = plt.subplots(2, 3, figsize=(12, 7))
plots = [
    (data["Age"], "Age - Original"), (result["Age_MinMax"], "Age - Min-Max"),
    (result["Age_Zscore"], "Age - Z-score"), (data["Marks"], "Marks - Original"),
    (result["Marks_MinMax"], "Marks - Min-Max"), (result["Marks_Zscore"], "Marks - Z-score")
]
for ax, (values, title) in zip(axes.flat, plots):
    ax.hist(values, bins=5, edgecolor="black")
    ax.set_title(title)
    ax.set_xlabel("Value")
    ax.set_ylabel("Count")
plt.tight_layout()
plt.show()
```

## Result and conclusion\n\nAge and Marks were scaled using Min-Max normalisation and Z-score standardisation. The first five values were compared and histograms were plotted before and after scaling.
