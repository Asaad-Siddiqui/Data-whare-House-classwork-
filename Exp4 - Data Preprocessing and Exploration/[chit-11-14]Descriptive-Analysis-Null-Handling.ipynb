## Step 1: Create or load the dataset\n\nWe use sample student data. To use your own CSV, replace the sample-data cell with `df = pd.read_csv('/content/your_file.csv')`.

> **What this cell does:** This creates a small student table. The `NaN` values are intentional: they represent missing information that we will find and fill later.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.DataFrame({
    "Name": ["Amit", "Sara", "Rahul", "Priya", "Zoya", "John", "Maya", "Ali"],
    "Age": [20, 21, np.nan, 22, 20, 23, 21, 24],
    "Marks": [78, 85, 90, np.nan, 65, 95, 72, 88],
    "Attendance": [85, 90, 88, 92, np.nan, 96, 80, 89],
    "Department": ["Computer", "IT", "Computer", "IT", "Computer", "IT", "Computer", "IT"]
})
display(df)
```

## Step 2: Display basic information\n\n`shape` shows rows and columns, `dtypes` shows data types, `head()` shows the first five rows, and `describe()` gives a summary.

> **What this cell does:** This shows the table size, column types, first five records, and a quick numeric summary. It helps us understand the dataset before changing it.


```python
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)
print("\nFirst five rows:")
display(df.head())
print("\nDescriptive statistics:")
display(df.describe())
```

## Step 3: Mean, median, mode, standard deviation and quartiles

> **What this cell does:** This calculates common summary values for each numeric column: mean, median, mode, standard deviation, and quartiles.


```python
numeric = df.select_dtypes(include="number")
summary = pd.DataFrame({
    "Mean": numeric.mean(),
    "Median": numeric.median(),
    "Mode": numeric.mode().iloc[0],
    "Standard Deviation": numeric.std(),
    "Q1 (25%)": numeric.quantile(0.25),
    "Q2 (50%)": numeric.quantile(0.50),
    "Q3 (75%)": numeric.quantile(0.75)
})
display(summary)
```

## Step 4: Find and fill null values\n\nFill numeric missing values with the mean and text missing values with the most common value (mode).

> **What this cell does:** First, we count missing values. Then we fill missing numbers with their column mean and missing text with the most common text value (mode). Finally, we check that no missing values remain.


```python
print("Null values before:")
print(df.isnull().sum())

clean_df = df.copy()
for col in clean_df.select_dtypes(include="number").columns:
    clean_df[col] = clean_df[col].fillna(clean_df[col].mean())
for col in clean_df.select_dtypes(exclude="number").columns:
    if clean_df[col].isnull().any():
        clean_df[col] = clean_df[col].fillna(clean_df[col].mode()[0])

print("\nNull values after:")
print(clean_df.isnull().sum())
display(clean_df)
```

## Result and conclusion\n\nThe data shape, data types, first rows and descriptive statistics were displayed. Null values were identified and filled using mean for numeric columns and mode for categorical columns.
