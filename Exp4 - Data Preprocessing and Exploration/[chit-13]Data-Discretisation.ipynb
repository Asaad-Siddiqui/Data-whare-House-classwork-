## Step 1: Create or load data\n\nDiscretisation changes numeric values into groups called bins. We use the Marks column.

> **What this cell does:** This creates sample student data. We will group the non-missing `Marks` values into three bins.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.DataFrame({
    "Name": ["Amit", "Sara", "Rahul", "Priya", "Zoya", "John", "Maya", "Ali"],
    "Age": [20, 21, np.nan, 22, 20, 23, 21, 24],
    "Marks": [78, 85, 90, np.nan, 65, 95, 72, 88],
    "Attendance": [85, 90, 88, 92, np.nan, 96, 80, 89],
    "Department": ["Computer", "IT", "Computer", "IT", "Computer", "IT", "Computer", "IT"]
})
display(df)
```

## Step 2: Equal-width binning\n\nEqual-width binning gives each bin the same numeric range. The number of values in each bin may differ.

> **What this cell does:** Equal-width binning divides the full marks range into three intervals of equal width. We display each interval and how many marks fall inside it.


```python
marks = df["Marks"].dropna()
bins_df = pd.DataFrame({"Marks": marks})
bins_df["Equal_Width_Bin"] = pd.cut(bins_df["Marks"], bins=3, include_lowest=True)

print("Equal-width bin counts:")
display(bins_df["Equal_Width_Bin"].value_counts(sort=False).rename("Count").to_frame())
display(bins_df.sort_values("Marks"))
```

## Step 3: Equal-frequency binning\n\nEqual-frequency binning tries to put a similar number of values into each bin.

> **What this cell does:** Equal-frequency binning uses quantiles to make bins with roughly similar numbers of records. The ranges may be different sizes. `duplicates='drop'` lets pandas handle repeated bin boundaries safely.


```python
bins_df["Equal_Frequency_Bin"] = pd.qcut(bins_df["Marks"], q=3, duplicates="drop")

print("Equal-frequency bin counts:")
display(bins_df["Equal_Frequency_Bin"].value_counts(sort=False).rename("Count").to_frame())
display(bins_df.sort_values("Marks"))
```

## Step 4: Difference between the methods\n\n- **Equal-width:** same numeric range per bin, but counts may differ.\n- **Equal-frequency:** similar number of values per bin, but ranges may differ.

> **What this cell does:** This prints the counts for both methods so you can compare them. Equal-width means equal range size; equal-frequency aims for similar numbers of records in each bin.


```python
print("Equal-width counts:")
display(bins_df["Equal_Width_Bin"].value_counts(sort=False))
print("Equal-frequency counts:")
display(bins_df["Equal_Frequency_Bin"].value_counts(sort=False))
```

## Result and conclusion\n\nThe Marks column was divided into three bins using equal-width and equal-frequency methods. The bin ranges and counts were displayed and compared.
