# CHIT No. 21 — K-means from Scratch (K = 2)

Implements K-means using only NumPy and Matplotlib—without sklearn KMeans. It uses the five given points, prints centroids after each iteration, and stops when assignments no longer change.

**How to use:** Run each cell from top to bottom. Read the short explanation before every code cell.

### What this cell does
Import NumPy for calculations and Matplotlib for the final plot.

```python
import numpy as np
import matplotlib.pyplot as plt
```

### What this cell does
Store the five points exactly as given in the question.

```python
points = np.array([
    [1, 1],
    [2, 1],
    [4, 3],
    [5, 4],
    [6, 5]
], dtype=float)

print(points)
```

### What this cell does
Choose two starting centroids from the given points. These are the initial centres for the two clusters.

```python
centroids = np.array([
    [1, 1],
    [5, 4]
], dtype=float)

print("Initial centroids:")
print(centroids)
```

### What this cell does
Run K-means manually: calculate distances, assign each point to the nearest centroid, and update each centroid to the mean of its assigned points. Print centroids every iteration and stop when cluster assignments stop changing.

```python
previous_labels = None
iteration = 1

while True:
    # Calculate distance from every point to every centroid
    distances = np.linalg.norm(points[:, np.newaxis] - centroids, axis=2)

    # Assign each point to the nearest centroid
    labels = np.argmin(distances, axis=1)

    # Update centroids using the mean of points in each cluster
    new_centroids = centroids.copy()
    for cluster_id in range(2):
        cluster_points = points[labels == cluster_id]
        if len(cluster_points) > 0:
            new_centroids[cluster_id] = cluster_points.mean(axis=0)

    print(f"Iteration {iteration} centroids:")
    print(new_centroids)

    converged = previous_labels is not None and np.array_equal(labels, previous_labels)
    centroids = new_centroids

    if converged:
        print("Converged: cluster assignments did not change.")
        break

    previous_labels = labels.copy()
    iteration += 1
```

### What this cell does
Print the final cluster membership for each point.

```python
for cluster_id in range(2):
    print(f"Cluster {cluster_id + 1}:")
    print(points[labels == cluster_id])
```

### What this cell does
Draw the final clusters and mark their centroids with X symbols.

```python
plt.figure(figsize=(7, 5))
for cluster_id in range(2):
    cluster_points = points[labels == cluster_id]
    plt.scatter(cluster_points[:, 0], cluster_points[:, 1], s=80,
                label=f"Cluster {cluster_id + 1}")

plt.scatter(centroids[:, 0], centroids[:, 1], marker="X", s=250,
            c="black", label="Centroids")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("K-means from Scratch (K = 2)")
plt.legend()
plt.grid(True)
plt.show()
```
