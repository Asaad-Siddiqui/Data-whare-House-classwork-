# CHIT No. 17 — Z-Score Outliers and Winsorisation

This notebook detects outliers where the absolute Z-score is greater than 3, caps extreme values, and compares summary statistics before and after treatment.

**How to use:** Run the cells from top to bottom. Each code cell has a short explanation before it.

### What this cell does
Import the libraries for data handling, Z-scores, and plotting.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore
```

### What this cell does
Create a numeric sample with one unusually large value.

```python
data = pd.DataFrame({
    "Salary": [25000, 27000, 28000, 29000, 30000, 31000, 32000, 33000, 34000, 35000, 200000]
})

data
```

### What this cell does
Calculate each value's Z-score. A Z-score tells how many standard deviations a value is from the mean.

```python
data["Z_Score"] = zscore(data["Salary"])
data
```

### What this cell does
Find rows where the absolute Z-score is greater than 3, as required in the question.

```python
outliers = data[data["Z_Score"].abs() > 3]

print("Outlier rows (|Z-score| > 3):")
print(outliers)
```

### What this cell does
Save the original salary values and cap values to the 5th and 95th percentiles. This is winsorisation/capping.

```python
data["Salary_Capped"] = data["Salary"].clip(
    lower=data["Salary"].quantile(0.05),
    upper=data["Salary"].quantile(0.95)
)

data
```

### What this cell does
Compare common summary statistics before and after capping.

```python
before_after = pd.DataFrame({
    "Before Treatment": data["Salary"].describe(),
    "After Treatment": data["Salary_Capped"].describe()
})

before_after
```

### What this cell does
Plot the salary distributions before and after capping for a quick visual comparison.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].boxplot(data["Salary"])
axes[0].set_title("Before Capping")
axes[0].set_ylabel("Salary")

axes[1].boxplot(data["Salary_Capped"])
axes[1].set_title("After Capping")
axes[1].set_ylabel("Salary")

plt.tight_layout()
plt.show()
```
