# CHIT No. 18 — One-Hot Encoding, Label Encoding and Correlation

This notebook encodes a categorical column using one-hot and label encoding, displays the transformed data, and plots a correlation heatmap for numeric columns.

**How to use:** Run the cells from top to bottom. Each code cell has a short explanation before it.

### What this cell does
Import pandas, seaborn, matplotlib, and LabelEncoder.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

### What this cell does
Create a small sample dataframe containing categories and numeric columns.

```python
data = pd.DataFrame({
    "Department": ["IT", "HR", "Sales", "IT", "HR", "Sales"],
    "Experience": [2, 5, 3, 4, 6, 1],
    "Salary": [30000, 40000, 35000, 45000, 42000, 28000]
})

data
```

### What this cell does
Use one-hot encoding to create a separate 0/1 column for each department.

```python
one_hot_data = pd.get_dummies(data, columns=["Department"], dtype=int)
one_hot_data
```

### What this cell does
Use label encoding to convert each department name into a number. The encoded values are category labels, not rankings.

```python
label_data = data.copy()
encoder = LabelEncoder()
label_data["Department_Label"] = encoder.fit_transform(label_data["Department"])

print("Label mapping:")
for category, number in zip(encoder.classes_, encoder.transform(encoder.classes_)):
    print(category, "=", number)

label_data
```

### What this cell does
Select numeric columns from the one-hot encoded dataframe and calculate their correlation matrix.

```python
numeric_data = one_hot_data.select_dtypes(include="number")
correlation = numeric_data.corr()

correlation
```

### What this cell does
Draw a heatmap to show correlations between the numeric columns.

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
