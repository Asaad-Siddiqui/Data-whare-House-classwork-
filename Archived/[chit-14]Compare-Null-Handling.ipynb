## Step 1: Create or load dataset\n\nWe use a dataset with missing values and compare three methods using the same original data.

> **What this cell does:** This creates student data with missing values. We will make three copies so each method starts from the same original data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.DataFrame({
    "Name": ["Amit", "Sara", "Rahul", "Priya", "Zoya", "John", "Maya", "Ali"],
    "Age": [20, 21, np.nan, 22, 20, 23, 21, 24],
    "Marks": [78, 85, 90, np.nan, 65, 95, 72, 88],
    "Attendance": [85, 90, 88, 92, np.nan, 96, 80, 89],
    "Department": ["Computer", "IT", "Computer", "IT", "Computer", "IT", "Computer", "IT"]
})
display(df)
```

## Step 2: Apply the three methods\n\n- **Drop rows:** remove rows containing missing values.\n- **Mean imputation:** fill missing numeric values with the column mean.\n- **Forward fill:** copy the previous row's value into a missing cell.

> **What this cell does:** We create three versions of the dataset: `drop_df` removes incomplete rows, `mean_df` fills missing values with the mean or mode, and `ffill_df` copies the previous row's value into missing cells. We then compare their sizes.


```python
drop_df = df.dropna().copy()

mean_df = df.copy()
for col in mean_df.select_dtypes(include="number").columns:
    mean_df[col] = mean_df[col].fillna(mean_df[col].mean())
for col in mean_df.select_dtypes(exclude="number").columns:
    if mean_df[col].isnull().any():
        mean_df[col] = mean_df[col].fillna(mean_df[col].mode()[0])

ffill_df = df.ffill()

print("Drop rows shape:", drop_df.shape)
print("Mean imputation shape:", mean_df.shape)
print("Forward fill shape:", ffill_df.shape)
```

## Step 3: Compare shape and mean of Marks

> **What this cell does:** This table compares the number of rows and columns and the average Marks value after each method. It shows how the choice of missing-value method can affect the dataset.


```python
comparison = pd.DataFrame({
    "Method": ["Original", "Drop rows", "Mean imputation", "Forward fill"],
    "Rows": [df.shape[0], drop_df.shape[0], mean_df.shape[0], ffill_df.shape[0]],
    "Columns": [df.shape[1], drop_df.shape[1], mean_df.shape[1], ffill_df.shape[1]],
    "Mean Marks": [
        df["Marks"].mean(), drop_df["Marks"].mean(),
        mean_df["Marks"].mean(), ffill_df["Marks"].mean()
    ]
})
display(comparison.round(2))
```

## Which method would you use?\n\nFor this sample, mean imputation is a simple choice because it keeps all rows and fills missing numeric values. Drop rows can be useful when only a few rows are incomplete. Forward fill is usually better for ordered data such as time series; it may not be suitable for unrelated student records.

## Result and conclusion\n\nDrop rows, mean imputation and forward fill were applied to the same data. Their shapes and mean Marks values were compared.
